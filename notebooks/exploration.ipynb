- Distribute classes across 5 workers, so that each worker has rougly the same number of classes
- Since classes for these dataset are roughly the same we can use a round robin strategy

```python
from collections import defaultdict

n_workers = 5
classes = sorted(df["intent"].unique())

worker_classes = defaultdict(list)

for index, class_name in enumerate(classes):
    worker_id = index % n_workers
    worker_classes[worker_id].append(class_name)

worker_classes
```
